### Setting up Langchain and OpenAI
**Goals**
- Setup Langchain, Langsmith and Langserve
- Use basic common components: prompt templates, models and output parsers
- Build a simple application with langchain
- Trace the application with Langsmith
- Serve the application with Langserve

Setting up the environment

```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACTING_V2"] = "true"
os.environ["LANGCHAIN_PROJECT"] = os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(model="gpt-4o")
llm
```

Giving input and getting response from LLM

```python
result = llm.invoke("How to prevent Intellectual property from being stolen by big AI companies to train their models?")
print(result)
```

```python
result = llm.invoke("What is a chatprompt template in langchain?")
print(result)
```

### Chatprompt templates:
A strcutured format used to define and manage converational prompts for language models. They help shape how interactions are conducted with an AI model

```python
from langchain_core.prompts import ChatPromptTemplate
prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system","You are a professional trainer specializing in high endurance training. Provide me answers based on the questions"),
        ("user", "{input}")
        
    ]
)
```

```python
chain = prompt_template|llm
response = chain.invoke({"input": "Suggest some workouts that build extreme endurance"})
response
```

**Output Parsing**
String output parser

```python
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
chain = prompt_template|llm|output_parser
response = chain.invoke({"input": "Suggest a workout for durability and longevity"})
print(response)
```
