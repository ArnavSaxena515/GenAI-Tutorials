IMDB Dataset of 50K Movie Reviews

```python
import pandas as pd
```

```python
data = "../IMDB Dataset.csv"
df = pd.read_csv(data)
df.head()
```

```python
df = df.sample(frac=0.1, random_state=1)
df.shape
```

Converting to lower

```python
df['review'] = df['review'].str.lower()
df.head()
```

Removing HTML tags with regular expressions

```python
import re
def remove_html_tags(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)

def remove_url(text):
    url_pattern = re.compile(r'http\S+|www\S+|https\S+', re.IGNORECASE)
    return url_pattern.sub(r'', text)
```

```python
text  = "<html><head><title>A Title</title></head><body><h1>Header</h1><p>This is a paragraph.</p></body></html>"
remove_html_tags(text)
```

```python
df["review"] = df["review"].apply(remove_html_tags)
df.head()
```

Remove Punctuation

```python
import string, time
exclude = string.punctuation
exclude
```

```python
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', exclude))
```

```python
txt = 'Text. With. Punch,,.'
print(remove_punctuation(txt))
```

```python
df['review'] = df['review'].apply(remove_punctuation)
df.head()
```

Handling Abbreviations

```python
chat_words = {
    'AFAIK':'As Far As I Know',
    'AFK':'Away From Keyboard',
    'ASAP':'As Soon As Possible',
    "FYI": "For Your Information",
    "ASAP": "As Soon As Possible",
    "BRB": "Be Right Back",
    "BTW": "By The Way",
    "OMG": "Oh My God",
    "IMO": "In My Opinion",
    "LOL": "Laugh Out Loud",
    "TTYL": "Talk To You Later",
    "GTG": "Got To Go",
    "TTYT": "Talk To You Tomorrow",
    "IDK": "I Don't Know",
    "TMI": "Too Much Information",
    "IMHO": "In My Humble Opinion",
    "ICYMI": "In Case You Missed It",
    "AFAIK": "As Far As I Know",
    "BTW": "By The Way",
    "FAQ": "Frequently Asked Questions",
    "TGIF": "Thank God It's Friday",
    "FYA": "For Your Action",
    "ICYMI": "In Case You Missed It",
}

```

```python
def abbreviation_conversion(text):
    words = text.split()
    new_text = []
    for word in words:
        
        if word.upper() in chat_words:
          
            new_text.append(chat_words[word.upper()])
        else:
            new_text.append(word)
    return " ".join(new_text)
```

```python
abbreviation_conversion("FYI I will be AFK for a while. ASAP I will get back to you. TTYL")
```

Removing Stopwords

```python
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')
```

```python
def remove_stopwords(text):
    stop_words = set(stopwords.words('english'))
    words = text.split()
    new_text = []
    for word in words:
        if word not in stop_words:
            new_text.append(word)
    return " ".join(new_text)
```

```python
remove_stopwords("This is a sample sentence, showing off the stop words filtration.")
```

```python
df['review'] = df['review'].apply(remove_stopwords)
df.head()
```

Remove Emojis (Uni-Code characters)

```python
def remove_emojis(text):
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
remove_emojis("This is a test 😊. This is only a test 🚀!")
```
