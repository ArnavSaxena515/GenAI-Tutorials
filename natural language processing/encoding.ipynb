Bag of Words

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({
    'text': [
        'I love programming! 😊',
        'Python is awesome Data is Awesome Python awesome! 🚀',
        'Data Science is the future! 📊',
        'Statistics is boring! 🎲',
        'Machine Learning is fascinating! 🤖',
        'Mathemtics is hard! 🧮'
    ], 'label': [1, 1, 1, 0, 1, 0]
})

df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer()
```

```python
X = vectorizer.fit_transform(df['text'])
print(vectorizer.vocabulary_)
print(X.toarray())
```

```python
vectorizer.transform(["I love Python Python is love! 😊"]).toarray()
```

N-grams: the N defines how many words to consider in a token

```python
vectorizer2 = CountVectorizer(ngram_range=(2,2))
```

```python
X2 = vectorizer2.fit_transform(df['text'])
print(vectorizer2.vocabulary_)
```

```python
print(X2.toarray())
```

TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
TfidfVectorizer = TfidfVectorizer()
```

```python
X3 = TfidfVectorizer.fit_transform(df['text'])
print(TfidfVectorizer.vocabulary_)
print(X3.toarray())
```
