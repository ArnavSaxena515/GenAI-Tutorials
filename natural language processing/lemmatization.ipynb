### Lemmatization
It is a natural language processing (NLP) technique used to reduce words to their base or dictionary form — known as a lemma. eg: running --> run, better --> good, was --> be

“better” → “good” is only possible with semantic understanding
It’s not just cutting characters — it’s using a dictionary and grammar rules

📚 How it works:
The word is checked against a lexicon (dictionary)

Part-of-speech tagging is used to understand context (noun, verb, etc.)

Then it's reduced to its lemma

✅ Why it's useful in NLP:
Improves text normalization for downstream tasks like search, classification, and topic modeling

Helps models understand that “run,” “runs,” “ran,” and “running” all relate to the same concept

**✅ Why Lemmatization is Better Than Stemming**
1. It produces real words:
Stemming can produce broken or nonsense words:

“Studies” → studi ❌

“Caring” → car ❌

Lemmatization gives proper base forms:

“Studies” → study ✅

“Caring” → care ✅

2. It understands grammar: 
Lemmatization uses part-of-speech tagging (noun, verb, adjective) to decide how to reduce the word.

Example:

"Better" → good (as an adjective) ✅

A stemmer can’t do this — it’s blind to grammar.

3. It improves meaning + context
Stemming just cuts — it doesn’t know that “ran” and “running” relate to “run”.

Lemmatization does.

So models trained on lemmatized text can better generalize.

4. Better for downstream NLP tasks
Text classification

Information retrieval (search engines)

Semantic analysis

All benefit from cleaner, more accurate base forms.

⚠️ When Stemming Might Be Better
Stemming isn’t useless — it’s just crude. It can be useful when:

You need super fast preprocessing (e.g. billions of documents)

You don’t care about perfect grammar (e.g. keyword matching)

But in most modern NLP tasks (especially with LLMs or vector search):

Lemmatization > Stemming


```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
'''
POS argument tells it what to treat the word as
POS - Noun - n
Verb - v
Adjective - a
adverb - r
'''

lemmatizer.lemmatize("going", pos="v")
```
